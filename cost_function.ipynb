```python
 x =np.array([2, 3, 4, 5])
 y =np.array([4, 7, 9, 12])
```

```python
def gradient(w):
  pred = w * x
  return np.mean(2 * x * (pred - y))

for w in [0, 1, 2, 3]:
  print(f"w={w}: gradient={gradient(w):.1f}")
```

TASK 7

```python
w = 0.0
lr = 0.02

prev_cost = cost(w)
step = 0

while True:
    g = gradient(w)
    w = w - lr * g
    new_cost = cost(w)
    step += 1
    print(f"{step}: w={w:.4f}, cost={new_cost:.4f}")
    if abs(prev_cost - new_cost) < 0.0001:
        break
    prev_cost = new_cost

print(f"\nStops in {step}th step. Final answer: w={w:.4f}, cost={new_cost:.4f}")
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [5, 8, 10, 15, 17, 20]
n = len(x)

def cost(w, b):
    return sum((w*x[i] + b - y[i])**2 for i in range(n))

def grad_w(w, b):
    return sum(2*x[i]*(w*x[i] + b - y[i]) for i in range(n))

def grad_b(w, b):
    return sum(2*(w*x[i] + b - y[i]) for i in range(n))

w, b = 0.0, 0.0
lr = 0.01
costs = []

for step in range(300):
    dw = grad_w(w, b)
    db = grad_b(w, b)
    w = w - lr * dw
    b = b - lr * db
    costs.append(cost(w, b))

print(f"Yakuniy w={w:.4f}, b={b:.4f}, cost={costs[-1]:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(costs)
plt.xlabel('Qadam')
plt.ylabel('Cost')
plt.title('Cost qadam-baqadam pasayishi')
plt.grid(True, alpha=0.3)
plt.show()
```
