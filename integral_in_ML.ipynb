 c) ∫(4x³ + 3x²) dx =x^4+x^3+c

```python
x=sp.symbols("x")
F1=sp.integrate(x**6,x)
print(f'∫x^6 dx= {F1} + C')

F2=sp.integrate(9,x)
print(f'∫9 dx= {F2} + C')

F3=sp.integrate((4*x**3+3*x**2),x)
print(f'∫(4*x^3+3*x^2) dx ={F3} + C')

F4=sp.integrate((2*x**5),x)
print(f'∫2*x^5 dx = {F4} + C')
```

```python
def f(x):
  return x**3

def left_riemann(f, a, b, n):
    dx = (b - a) / n
    xs = np.linspace(a, b, n, endpoint=False)
    return np.sum(f(xs)) * dx

for n in [5,20]:
  print(f'n={n}:  {left_riemann(f,0,2,n):.2f} ')
```

```python
xs = np.linspace(0, 2,6)
ys = f(xs)
natija = np.trapezoid(ys, xs)
print(f'trapetsiya (n=5) :natija: {natija}')
```
