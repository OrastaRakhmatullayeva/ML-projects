```python
#∇f(1, 2)
#f(x, y) = 3x^2 + 2xy - y^2
def gradient(f, x, y, h=1e-5): #0.000001
    gx = (f(x+h, y) - f(x, y)) / h  # x bo'yicha
    gy = (f(x, y+h) - f(x, y)) / h  # y bo'yicha
    return np.array([gx, gy])

def f(x, y): return 3*x**2 + 2*x*y-y**2

print(gradient(f, 1, 2))
```
