TASK 1"

1) Funksiya: y = (5x + 2)³
2) Ichki funksiya g(x) va tashqi funksiya f(u) ni ayting
3) x = 1, 2, 3 uchun avval u = g(x) ni, keyin y = f(u) ni bosqichma-bosqich hisoblang.
4) Natijalarni y(x) = f(g(x)) funksiyasi sifatida to'g'ridan-to'g'ri hisoblab tekshiring.


```python
def g(x):
    return 5*x + 2

def f(u):
    return u**3

def y1(x):
    u = g(x)
    y = f(u)
    return u, y

for x in [1, 2, 3]:
    u, y = y1(x)
    print(f"x={x}: u={u}, y={y}")
```

TASK 2

 Funksiya: y = (x − 4)²

2) Ichki va tashqi funksiyalarni ajratib, chain rule bilan dy/dx formulasini QO'LDA toping

3) x = 0, 4, 6 nuqtalarida dy/dx qiymatlarini hisoblang.

4) sp.diff bilan tekshiring.


```python
import sympy as sp

x = sp.Symbol('x')
y = (x-4)**2
Y=sp.diff(y, x)
print(Y)

points = [0, 4, 6]
for p in points:
    print(f"x={p}, function:{Y.subs(x, p)}")
```

```python
# y = (x − 4)²
# g(x)=x-4      f(u)=u^2
x=sp.symbols("x")

f=x**2-8*x+16
F=sp.diff(f)
print(F)
print(F.subs(x,4))
```

3) Chain rule bilan dy/dx formulasini QO'LDA chiqaring.

```python
# y = (x² − 2x)³
def g(x):
  return x**2-2*x

def f(u):
  return u**3

def y1(x):
  u=g(x)
  y=f(u)
  return u,y

print(y1(x))

for x in [0,1,3]:
    u, y = y1(x)
    print(f"x={x}: u={u}, y={y}")
```

2) Dars 26, 14-slayddagi numerik shablon asosida (h=0.0001) dy/dx ni taxminlang.
3) Keyin chain rule bilan QO'LDA analitik dy/dx formulasini toping va x=1 da hisoblang.


```python
def g(x): return 2*x + 1
def f(u): return u ** 4
def y(x): return f(g(x))  # murakkab funksiya

h = 0.0001
x0 = 1

numeric_dydx = (y(x0+h) - y(x0)) / h
chain_rule_dydx = 4 * g(x0)**3 * (2)  # dy/du · du/dx

print(f"Numerik: {numeric_dydx:.3f}")      # ≈ 42.001
print(f"Chain rule: {chain_rule_dydx}")    # 42
```
